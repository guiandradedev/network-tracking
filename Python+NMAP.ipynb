O Python não realiza a varredura de rede sozinho. Para facilitar a integração com o Nmap, existe a biblioteca python-nmap, que funciona como uma interface entre o código Python e o executável do Nmap instalado na máquina.

Essa biblioteca torna o desenvolvimento mais simples, pois os resultados das varreduras já são retornados em estruturas Python, como dicionários, evitando a necessidade de interpretar manualmente a saída do terminal.

É importante observar que a biblioteca não substitui o Nmap. Ela apenas envia comandos ao Nmap e recebe os resultados processados. Portanto, antes de executar o código, o Nmap precisa estar instalado e acessível pelo sistema operacional.

```python
import nmap

scanner = nmap.PortScanner()

scanner.scan(
    hosts='localhost',
    arguments='-sV -sC'
)

for host in scanner.all_hosts():

    print(f'Host: {host}')
    print(f'Status: {scanner[host].state()}')

    for protocolo in scanner[host].all_protocols():

        print(f'\nProtocolo: {protocolo}')

        portas = scanner[host][protocolo].keys()

        for porta in portas:

            servico = scanner[host][protocolo][porta]

            print(
                f'Porta: {porta} | '
                f'Estado: {servico["state"]} | '
                f'Serviço: {servico["name"]}'
            )
```
